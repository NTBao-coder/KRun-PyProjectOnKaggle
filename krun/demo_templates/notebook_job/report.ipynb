# Batch notebook demo
Summarize a few values and save an artifact.

```python
from pathlib import Path
import json
values = [2, 4, 6, 8]
result = {'count': len(values), 'sum': sum(values)}
print(result)
```

```python
%time total = sum(values)
Path('outputs').mkdir(exist_ok=True)
Path('outputs/result.json').write_text(json.dumps(result) + '\n', encoding='utf-8')
```
